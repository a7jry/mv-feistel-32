```python
from sage.all import *
import random
from itertools import product


def random_boolean_poly(B, p_lin=0.65, p_quad=0.30, p_const=0.5):
    """Random polynomial with linear, quadratic and constant terms."""
    x = B.gens()
    n = len(x)
    terms = [x[i] for i in range(n) if random.random() < p_lin]
    terms += [x[i] * x[j]
              for i in range(n) for j in range(i + 1, n)
              if random.random() < p_quad]
    if random.random() < p_const:
        terms.append(B.one())
    return sum(terms, B.zero())


def find_system(B, n, max_attempts=None):
    """
    Generate n random polynomials in n variables until the reduced
    Groebner basis of the Boolean ideal (field equations are implicit
    in BooleanPolynomialRing) has exactly n elements. Return the GB.
    """
    attempt = 0
    while max_attempts is None or attempt < max_attempts:
        attempt += 1
        eqs = [random_boolean_poly(B) for _ in range(n)]
        gb = B.ideal(eqs).groebner_basis()
        if len(gb) == n:
            return list(gb)
    raise RuntimeError(f"No system with GB of size {n} found "
                       f"in {max_attempts} attempts.")


def create_composed_system(n, s, max_attempts=None, seed=None):
    """
    Build s random systems (n polys, n vars, GB of size n), then
    shuffle each system's polynomial order and compose them:
        F = G_s o shuffle(G_{s-1}) o ... o shuffle(G_1)
    """
    if seed is not None:
        random.seed(seed)
        set_random_seed(seed)

    B = BooleanPolynomialRing(n, [f"x{i}" for i in range(n)],
                              order="degrevlex")

    current = find_system(B, n, max_attempts)
    for step in range(2, s + 1):
        random.shuffle(current)                      # shuffle polynomial order
        nxt = find_system(B, n, max_attempts)
        # substitute variable x_i of the next system by current[i]
        current = [g(*current) for g in nxt]
        print(f"step {step}/{s}: max degree = "
              f"{max(p.degree() for p in current)}")
    return current


def report_output_bias(system):
    """Fraction of zero/one outputs of each polynomial over all inputs."""
    n = system[0].parent().ngens()
    inputs = list(product([0, 1], repeat=n))
    for i, p in enumerate(system):
        ones = sum(int(p(*bits)) for bits in inputs)
        print(f"f_{i}: ones={ones}/{len(inputs)}, "
              f"zeros={len(inputs) - ones}/{len(inputs)}")


if __name__ == "__main__":
    system = create_composed_system(n=6, s=3, seed=1)
    for p in system:
        print(p)
    report_output_bias(system)
```

```python
def test_system(system, input_bits):
    if len(input_bits) != 8 or any(b not in "01" for b in input_bits):
        raise ValueError("Input must be exactly 8 bits.")

    # Convert input bits to GF(2) elements
    values = [GF(2)(int(b)) for b in input_bits]

    # Substitute the input into every polynomial
    output_bits = []
    for poly in system:
        value = poly(*values)
        output_bits.append(str(int(value)))

    output = ''.join(output_bits)

    print("Input : ", input_bits)
    print("Output:", output)

    return output
    
```

```python
import string

# Printable ASCII characters: space through ~
PRINTABLE = ''.join(chr(i) for i in range(32, 127))


def encrypt_text(system, text):
    # Convert plaintext to 8-bit ASCII
    plaintext_bits = ''.join(format(ord(c), '08b') for c in text)

    # Make length a multiple of 8 bits
    padding = (8 - len(plaintext_bits) % 8) % 8
    plaintext_bits += '0' * padding

    encrypted_bits = ""

    # Process 8 bits at a time
    for i in range(0, len(plaintext_bits), 8):
        block = plaintext_bits[i:i+8]

        # Convert bits to GF(2)
        values = [GF(2)(int(b)) for b in block]

        # Evaluate the polynomial functions
        output = ''.join(
            str(int(poly(*values)))
            for poly in system
        )

        encrypted_bits += output

    # Convert encrypted bits into integers, 8 bits at a time
    encrypted_values = [
        int(encrypted_bits[i:i+8], 2)
        for i in range(0, len(encrypted_bits), 8)
    ]

    # Map each 8-bit value to a printable ASCII character
    encrypted_text = ""

    for value in encrypted_values:
        encrypted_text += PRINTABLE[value % len(PRINTABLE)]

    print("Plaintext:        ", text)
    print("Plaintext binary: ", plaintext_bits)
    print("Encrypted binary: ", encrypted_bits)
    print("Encrypted text:   ", encrypted_text)

    return encrypted_text
```

```python

from sage.all import GF

# ============================================
# 1. XOR
# ============================================

def xor_bits(a, b):
    if len(a) != len(b):
        raise ValueError("Inputs must have the same length.")

    return ''.join(
        '0' if x == y else '1'
        for x, y in zip(a, b)
    )


# ============================================
# 2. 8-bit Polynomial System
#    -> 8-bit output
# ============================================

def evaluate_8bit_system(system, bits):
    if len(bits) != 8 or any(b not in "01" for b in bits):
        raise ValueError("Input must be exactly 8 bits.")

    if len(system) != 8:
        raise ValueError("System must contain exactly 8 polynomials.")

    values = [GF(2)(int(b)) for b in bits]

    return ''.join(
        str(int(poly(*values)))
        for poly in system
    )


# ============================================
# 3. Feistel Round Function
#    16-bit input -> 16-bit output
# ============================================

def round_function(system, bits):
    if len(bits) != 16:
        raise ValueError("Round input must be 16 bits.")

    left_byte = bits[:8]
    right_byte = bits[8:]

    left_output = evaluate_8bit_system(system, left_byte)
    right_output = evaluate_8bit_system(system, right_byte)

    return left_output + right_output


# ============================================
# 4. Feistel Encryption
#    32-bit block
# ============================================

def feistel_encrypt(system, plaintext, rounds=8):
    if len(plaintext) != 32 or any(b not in "01" for b in plaintext):
        raise ValueError("Plaintext must be exactly 32 bits.")

    L = plaintext[:16]
    R = plaintext[16:]

    for _ in range(rounds):
        F = round_function(system, R)

        L, R = R, xor_bits(L, F)

    return L + R


# ============================================
# 5. Feistel Decryption
# ============================================

def feistel_decrypt(system, ciphertext, rounds=8):
    if len(ciphertext) != 32 or any(b not in "01" for b in ciphertext):
        raise ValueError("Ciphertext must be exactly 32 bits.")

    L = ciphertext[:16]
    R = ciphertext[16:]

    for _ in range(rounds):
        # Invert one Feistel round
        old_R = L
        old_L = xor_bits(R, round_function(system, old_R))

        L, R = old_L, old_R

    return L + R


# ============================================
# 6. Hexadecimal <-> Binary
# ============================================

def hex_to_bits(value):
    if len(value) != 8:
        raise ValueError("Enter exactly 8 hexadecimal characters.")

    try:
        number = int(value, 16)
    except ValueError:
        raise ValueError("Use only hexadecimal characters: 0-9, A-F.")

    return format(number, '032b')


def bits_to_hex(bits):
    if len(bits) != 32:
        raise ValueError("Input must be exactly 32 bits.")

    return format(int(bits, 2), '08X')


# ============================================
# 7. Encrypt / Decrypt Hexadecimal
# ============================================

def encrypt_hex(system, plaintext_hex, rounds=8):
    bits = hex_to_bits(plaintext_hex.strip())
    ciphertext = feistel_encrypt(system, bits, rounds)
    return bits_to_hex(ciphertext)


def decrypt_hex(system, ciphertext_hex, rounds=8):
    bits = hex_to_bits(ciphertext_hex.strip())
    plaintext = feistel_decrypt(system, bits, rounds)
    return bits_to_hex(plaintext)


# ============================================
# 8. Interactive Test
# ============================================

def test_cipher(system, rounds=8):
    while True:
        print("\n--- 32-bit Feistel Cipher ---")
        print("1. Encrypt")
        print("2. Decrypt")
        print("3. Exit")

        choice = input("Choose: ").strip()

        if choice == "1":
            plaintext = input("Enter 8-digit hex plaintext: ").strip().upper()

            try:
                ciphertext = encrypt_hex(system, plaintext, rounds)
                print("Ciphertext:", ciphertext)
            except ValueError as e:
                print("Error:", e)

        elif choice == "2":
            ciphertext = input("Enter 8-digit hex ciphertext: ").strip().upper()

            try:
                plaintext = decrypt_hex(system, ciphertext, rounds)
                print("Decrypted:", plaintext)
            except ValueError as e:
                print("Error:", e)

        elif choice == "3":
            break

        else:
            print("Choose 1, 2, or 3.")

```

```python

# ============================================
# Arbitrary-Length Text Encryption
# Uses the existing 32-bit Feistel cipher
# ============================================

def encrypt_text(system, text, rounds=8):
    # Convert any UTF-8 text to bytes
    data = text.encode("utf-8")

    # PKCS#7-style padding to a multiple of 4 bytes
    padding_length = 4 - (len(data) % 4)
    data += bytes([padding_length]) * padding_length

    ciphertext = bytearray()

    # Encrypt each 4-byte (32-bit) block
    for i in range(0, len(data), 4):
        block = data[i:i + 4]
        bits = ''.join(format(byte, "08b") for byte in block)

        encrypted_bits = feistel_encrypt(
            system, bits, rounds
        )

        ciphertext.extend(
            int(encrypted_bits, 2).to_bytes(4, "big")
        )

    # Hex is a safe printable representation of ciphertext bytes
    return ciphertext.hex().upper()


def decrypt_text(system, ciphertext_hex, rounds=8):
    # Validate hexadecimal input
    ciphertext_hex = ciphertext_hex.strip()

    try:
        ciphertext = bytes.fromhex(ciphertext_hex)
    except ValueError:
        raise ValueError("Ciphertext must be valid hexadecimal.")

    if not ciphertext or len(ciphertext) % 4 != 0:
        raise ValueError(
            "Ciphertext must contain a positive multiple of 4 bytes."
        )

    plaintext = bytearray()

    # Decrypt each 4-byte (32-bit) block
    for i in range(0, len(ciphertext), 4):
        block = ciphertext[i:i + 4]
        bits = ''.join(format(byte, "08b") for byte in block)

        decrypted_bits = feistel_decrypt(
            system, bits, rounds
        )

        plaintext.extend(
            int(decrypted_bits, 2).to_bytes(4, "big")
        )

    # Validate and remove padding
    if not plaintext:
        raise ValueError("Decryption produced no data.")

    padding_length = plaintext[-1]

    if not 1 <= padding_length <= 4:
        raise ValueError("Invalid padding; decryption may have failed.")

    if plaintext[-padding_length:] != bytes([padding_length]) * padding_length:
        raise ValueError("Invalid padding; decryption may have failed.")

    plaintext = plaintext[:-padding_length]

    return plaintext.decode("utf-8")


```

```python

def test_cipher(system, text, rounds=8):
    # Plaintext representations
    plaintext_bytes = text.encode("utf-8")
    plaintext_hex = plaintext_bytes.hex().upper()
    plaintext_binary = ''.join(
        format(byte, '08b') for byte in plaintext_bytes
    )

    # Encrypt
    ciphertext_hex = encrypt_text(system, text, rounds)
    ciphertext_bytes = bytes.fromhex(ciphertext_hex)

    ciphertext_binary = ''.join(
        format(byte, '08b') for byte in ciphertext_bytes
    )

    # Display plaintext
    print("========== PLAINTEXT ==========")
    print("Text   :", repr(text))
    print("Hex    :", plaintext_hex)
    print("Binary :", plaintext_binary)

    # Display ciphertext
    print("\n========== CIPHERTEXT ==========")
    print("Text   :", repr(ciphertext_bytes.decode(
        "utf-8", errors="replace"
    )))
    print("Hex    :", ciphertext_hex)
    print("Binary :", ciphertext_binary)

    # Decrypt and verify
    recovered = decrypt_text(system, ciphertext_hex, rounds)

    print("\n========== VERIFICATION ==========")
    print("Decrypted:", repr(recovered))
    print("Test passed:", recovered == text)

    return ciphertext_hex
```
